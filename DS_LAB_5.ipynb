```python
# SIVAJI PRASHANTH
# 23102A010375
# CSE(AIML)
# EXP - 5
# Data Visualization with Matplotlib and Seaborn

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
import os

# ✅ Ensure output folder exists
output_dir = "/content/plots"
os.makedirs(output_dir, exist_ok=True)

# ---------------------------
# A. Line Plot
# ---------------------------
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y, marker='o', linestyle='--', color='b')
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.savefig(f"{output_dir}/line_plot.png")
plt.show()
plt.close()

# ---------------------------
# B. Bar Plot
# ---------------------------
df = pd.DataFrame({'Category': ['A', 'B', 'C'], 'Values': [10, 20, 15]})
df.plot(kind='bar', x='Category', y='Values', color=['red', 'green', 'blue'])
plt.title('Bar Plot')
plt.savefig(f"{output_dir}/bar_plot.png")
plt.show()
plt.close()

# ---------------------------
# C. Histogram and Density Plot
# ---------------------------
data = np.random.randn(1000)
sns.histplot(data, bins=30, kde=True, color='purple')
plt.title('Histogram and Density')
plt.savefig(f"{output_dir}/histogram_density.png")
plt.show()
plt.close()

# ---------------------------
# D. Scatter Plot
# ---------------------------
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, c='orange', s=80, edgecolors='black')
plt.title('Scatter Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.savefig(f"{output_dir}/scatter_plot.png")
plt.show()
plt.close()

# ---------------------------
# E. Box Plot
# ---------------------------
sns.boxplot(data=np.random.randn(100, 4))
plt.title('Box Plot')
plt.xticks([0, 1, 2, 3], ['Feature1', 'Feature2', 'Feature3', 'Feature4'])
plt.savefig(f"{output_dir}/box_plot.png")
plt.show()
plt.close()

print('All visualizations generated successfully and saved in /content/plots/')
```
